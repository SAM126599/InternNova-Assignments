```python
# Task 7: Handling Missing Values (10 Marks)

print("\n" + "=" * 80)
print("TASK 7: Handling Missing Values")
print("=" * 80)

# Create a dataset with missing values
data_missing = {
    'Student_ID': [1, 2, 3, 4, 5, 6, 7],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank', 'Grace'],
    'Math': [85, 92, np.nan, 90, 88, 76, 95],
    'Physics': [78, np.nan, 85, 92, np.nan, 88, 90],
    'Chemistry': [90, 88, 75, np.nan, 82, 95, 88],
    'Attendance': [95, 88, 92, 85, np.nan, 90, 87]
}

df_missing = pd.DataFrame(data_missing)

print("\nDataset with Missing Values:")
print(df_missing)

# Identify missing values using isnull() / isna()
print("\nMissing Values (isnull()):")
print(df_missing.isnull())

# Count missing values in each column
print("\nMissing Values Count per Column:")
print(df_missing.isnull().sum())

# Remove rows containing missing values
df_dropped = df_missing.dropna()
print("\nDataset after dropping rows with missing values:")
print(df_dropped)

# Fill missing values using appropriate methods
df_filled = df_missing.copy()
df_filled['Math'] = df_filled['Math'].fillna(df_filled['Math'].mean())
df_filled['Physics'] = df_filled['Physics'].fillna(df_filled['Physics'].median())
df_filled['Chemistry'] = df_filled['Chemistry'].fillna(0)  # Fill with 0
df_filled['Attendance'] = df_filled['Attendance'].fillna(df_filled['Attendance'].mode()[0])

print("\nDataset after filling missing values:")
print(df_filled)

print("\n" + "-" * 80)
print("Why handling missing data is important in Data Analytics:")
print("-" * 80)
print("""
1. Data Quality: Missing values can lead to biased or inaccurate analysis results.
2. Statistical Validity: Many statistical methods require complete datasets to produce valid results.
3. Machine Learning: Most ML algorithms cannot handle missing values and will fail or produce poor results.
4. Decision Making: Incomplete data can lead to incorrect business decisions and insights.
5. Data Integrity: Handling missing values properly maintains the integrity and reliability of your analysis.
6. Pattern Recognition: Understanding why data is missing can reveal important patterns about data collection.
""")
```

```python
# Task 8: Merge, Concatenate, GroupBy & Pivot Table (15 Marks)

print("\n" + "=" * 80)
print("TASK 8: Merge, Concatenate, GroupBy & Pivot Table")
print("=" * 80)

# Create two datasets for merging
df1 = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR']
})

df2 = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104, 105],
    'Salary': [75000, 65000, 80000, 70000, 68000],
    'Experience': [5, 3, 7, 4, 6]
})

print("\nDataFrame 1 (Employee Info):")
print(df1)
print("\nDataFrame 2 (Salary & Experience):")
print(df2)

# Merge two DataFrames using a common column
df_merged = pd.merge(df1, df2, on='Employee_ID')
print("\nMerged DataFrame (on Employee_ID):")
print(df_merged)

# Concatenate two DataFrames
df3 = pd.DataFrame({
    'Employee_ID': [106, 107],
    'Name': ['Frank', 'Grace'],
    'Department': ['IT', 'Finance']
})

df_concat = pd.concat([df1, df3], ignore_index=True)
print("\nConcatenated DataFrame:")
print(df_concat)

# GroupBy - Group data based on a selected column
print("\nGroupBy Department - Calculate aggregate values:")
grouped = df_merged.groupby('Department')

print("\nSum by Department:")
print(grouped['Salary'].sum())

print("\nMean by Department:")
print(grouped['Salary'].mean())

print("\nCount by Department:")
print(grouped['Salary'].count())

print("\nMin by Department:")
print(grouped['Salary'].min())

print("\nMax by Department:")
print(grouped['Salary'].max())

# Pivot Table
df_sales = pd.DataFrame({
    'Date': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02', '2024-01-03', '2024-01-03'],
    'Region': ['North', 'South', 'North', 'South', 'North', 'South'],
    'Product': ['A', 'A', 'B', 'B', 'A', 'B'],
    'Sales': [1000, 1500, 1200, 1800, 1100, 1600]
})

print("\nSales DataFrame:")
print(df_sales)

# Create a Pivot Table using Pandas
pivot = pd.pivot_table(df_sales,
                       values='Sales',
                       index='Region',
                       columns='Product',
                       aggfunc='sum',
                       margins=True,
                       margins_name='Total')

print("\nPivot Table (Sales by Region and Product):")
print(pivot)

```

```python
# Task 9: Exporting Data (5 Marks)

print("\n" + "=" * 80)
print("TASK 9: Exporting Data")
print("=" * 80)

# Export the final DataFrame to a new CSV file
df_merged.to_csv('processed_employees.csv', index=False)
print("\nExported merged DataFrame to: processed_employees.csv")

# Verify that the exported file contains the processed data
df_verify = pd.read_csv('processed_employees.csv')
print("\nVerification - Read back the exported file:")
print(df_verify)
```

```python
# Task 10: Mini Data Analysis Project (10 Marks)

print("\n" + "=" * 80)
print("TASK 10: Mini Data Analysis Project - Sales Dataset Analysis")
print("=" * 80)

# Create a comprehensive sales dataset
sales_data = {
    'Order_ID': [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010,
                 1011, 1012, 1013, 1014, 1015, 1016, 1017, 1018, 1019, 1020],
    'Date': ['2024-01-15', '2024-01-16', '2024-01-17', '2024-01-18', '2024-01-19',
             '2024-01-20', '2024-01-21', '2024-01-22', '2024-01-23', '2024-01-24',
             '2024-01-25', '2024-01-26', '2024-01-27', '2024-01-28', '2024-01-29',
             '2024-01-30', '2024-01-31', '2024-02-01', '2024-02-02', '2024-02-03'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Laptop', 'Mouse', 'Keyboard', 'Monitor',
                'Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Laptop', 'Mouse', 'Keyboard', 'Monitor',
                'Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Category': ['Electronics', 'Accessories', 'Accessories', 'Electronics', 'Electronics',
                 'Accessories', 'Accessories', 'Electronics', 'Electronics', 'Accessories',
                 'Accessories', 'Electronics', 'Electronics', 'Accessories', 'Accessories',
                 'Electronics', 'Electronics', 'Accessories', 'Accessories', 'Electronics'],
    'Quantity': [2, 15, 10, 5, 3, 20, 12, 6, 1, 25, 8, 4, 4, 18, 15, 7, 2, 22, 9, 5],
    'Unit_Price': [75000, 1200, 3500, 15000, 75000, 1200, 3500, 15000, 75000, 1200,
                   3500, 15000, 75000, 1200, 3500, 15000, 75000, 1200, 3500, 15000],
    'Customer_ID': ['C001', 'C002', 'C003', 'C004', 'C005', 'C006', 'C007', 'C008',
                    'C009', 'C010', 'C011', 'C012', 'C013', 'C014', 'C015', 'C016',
                    'C017', 'C018', 'C019', 'C020']
}

# Introduce some missing values for realistic scenario
sales_data['Quantity'][5] = np.nan
sales_data['Unit_Price'][12] = np.nan

df_sales_analysis = pd.DataFrame(sales_data)

print("\n1. Loading the Dataset:")
print("Dataset loaded with", len(df_sales_analysis), "rows and", len(df_sales_analysis.columns), "columns")

print("\n2. Data Inspection:")
print("\nFirst 5 rows:")
print(df_sales_analysis.head())

print("\nDataset Info:")
print(df_sales_analysis.info())

print("\n3. Identifying and Handling Missing Values:")
print("\nMissing values per column:")
print(df_sales_analysis.isnull().sum())

# Fill missing values
df_sales_analysis['Quantity'] = df_sales_analysis['Quantity'].fillna(df_sales_analysis['Quantity'].mean())
df_sales_analysis['Unit_Price'] = df_sales_analysis['Unit_Price'].fillna(df_sales_analysis['Unit_Price'].mean())

print("\nAfter filling missing values:")
print(df_sales_analysis.isnull().sum())

# Calculate Total Sales
df_sales_analysis['Total_Sales'] = df_sales_analysis['Quantity'] * df_sales_analysis['Unit_Price']

print("\n4. Selecting and Filtering Data:")
print("\nSelected columns (Order_ID, Product, Total_Sales):")
print(df_sales_analysis[['Order_ID', 'Product', 'Total_Sales']].head())

print("\nFilter: Electronics category with Total_Sales > 50000:")
print(df_sales_analysis[(df_sales_analysis['Category'] == 'Electronics') &
                        (df_sales_analysis['Total_Sales'] > 50000)])

print("\n5. Sorting Data:")
print("\nSorted by Total_Sales (Descending):")
print(df_sales_analysis.sort_values('Total_Sales', ascending=False).head())

print("\n6. Performing GroupBy Analysis:")
print("\nGroupBy Product - Total Sales:")
print(df_sales_analysis.groupby('Product')['Total_Sales'].sum())

print("\nGroupBy Region - Total Sales:")
print(df_sales_analysis.groupby('Region')['Total_Sales'].sum())

print("\nGroupBy Category - Total Sales and Average Quantity:")
print(df_sales_analysis.groupby('Category').agg({
    'Total_Sales': 'sum',
    'Quantity': 'mean'
}))

print("\n7. Creating Pivot Table:")
pivot_sales = pd.pivot_table(df_sales_analysis,
                             values='Total_Sales',
                             index='Region',
                             columns='Product',
                             aggfunc='sum',
                             margins=True,
                             margins_name='Total')

print("\nPivot Table - Total Sales by Region and Product:")
print(pivot_sales)

print("\n8. Finding Useful Insights from the Dataset:")
print("-" * 80)
print("Key Insights:")
print(f"1. Total Revenue: ₹{df_sales_analysis['Total_Sales'].sum():,.2f}")
print(f"2. Average Order Value: ₹{df_sales_analysis['Total_Sales'].mean():,.2f}")
print(f"3. Top Selling Product: {df_sales_analysis.groupby('Product')['Total_Sales'].sum().idxmax()}")
print(f"4. Top Performing Region: {df_sales_analysis.groupby('Region')['Total_Sales'].sum().idxmax()}")
print(f"5. Electronics vs Accessories Revenue:")
for category in df_sales_analysis['Category'].unique():
    revenue = df_sales_analysis[df_sales_analysis['Category'] == category]['Total_Sales'].sum()
    print(f"   - {category}: ₹{revenue:,.2f}")
print(f"6. Total Orders Processed: {len(df_sales_analysis)}")
print(f"7. Average Quantity per Order: {df_sales_analysis['Quantity'].mean():.2f}")

print("\n9. Exporting the Cleaned/Processed Dataset:")
df_sales_analysis.to_csv('processed_sales_analysis.csv', index=False)
print("Exported to: processed_sales_analysis.csv")

# Verify the exported file
df_verify_final = pd.read_csv('processed_sales_analysis.csv')
print(f"\nVerification - Exported file contains {len(df_verify_final)} rows")

print("\n" + "=" * 80)
print("ALL TASKS COMPLETED SUCCESSFULLY!")
print("=" * 80)
print("\nFiles created:")
print("1. sample_products.csv - Sample product data")
print("2. processed_employees.csv - Merged employee data")
print("3. processed_sales_analysis.csv - Final sales analysis dataset")
print("\nNote: All Python code is in this file: numpy_pandas_assignment.py")
```
